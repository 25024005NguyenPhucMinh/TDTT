Yêu cầu
Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14.
Input
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
Dòng 2: tham số tương ứng.
Output
In ra diện tích với 2 chữ số thập phân.

```python
class Shape:
    def area(self):
        return 0
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r
shapes = input()
if shapes == 'rectangle':
    w, h = map(float, input().split())
    shape = Rectangle(w, h)
elif shapes == 'circle':
    r = float(input())
    shape = Circle(r)
print(f'{shape.area():.2f}')
```

Xây dựng lớp Vehicle(brand, speed).

Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.

Viết phương thức info() cho mỗi lớp:
- Vehicle: brand-speed

Input

Dòng 1: loại đối tượng (Vehicle hoặc Car)

Dòng 2: dữ liệu tương ứng.

In ra chuỗi info(). 

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def info(self):
        return f'{self.brand} - {self.speed} km/h'
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
    def info(self):
        return f'{self.brand} - {self.speed} km/h - {self.seats} seats'
obj = input()
if obj == 'vehicle':
    brand, speed = input().split()
    obj = Vehicle(brand, speed)
elif obj == 'car':
    brand, speed, seats = input().split()
    obj = Car(brand, speed, seats)
print(obj.info())

```

Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance().

Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức

add_interest() để cộng balance*rate vào số dư.

Input

balance rate

Output

Số dư sau khi cộng lãi, in 2 chữ số thập phân.

```python
class Accout:
    def __init__(balance):
        self.balance = balance
    def deposit(self, x):
        self.balance += x
    def get_balance(self):
        return self.balance
class SavingsAccount(Accout):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
    def add_interest(self):
        self.balance += self.balance * self.rate
balance, rate = map(float, input().split())
acc = SavingsAccount(balance, rate)
acc.add_interest()
print(f'{acc.get_balance():.2f}')
```

Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:

[a b]

[c d]

Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.

Input

Một dòng chứa 4 số nguyên a b c d.

Output

In ra trace và determinant trên cùng một dòng. 

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def trace(self):
        return self.a + self.d
    def determinant(self):
        return self.a * self.d - self.b * self.c
a, b, c, d = map(int, input().split())
matrix = Matrix2x2(a, b, c, d)
print(matrix.trace(), matrix.determinant())
```

Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất:

P(x) = ax + b, Q(x) = cx + d.

Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.

Input

a b c d
Output

Chuỗi của đa thức P(x)+Q(x)

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        return f'{k}x + {m}'
a, b, c, d = map(float, input().split())
p_sum = PolynomialSum(a, b, c, d)
print(p_sum.sum_str())
```

Xây dựng lớp Product(name, price).

Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức

final_price() để trả về giá sau giảm.

Công thức: price * (100 - percent) / 100

Input

name price percent

Output

In ra giá cuối cùng với 2 chữ số thập phân. 


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class DiscountProduct(Product):

    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent

    def final_price(self):
        return self.price * (100 - self.percent) / 100
    
name, price, percent = input().split()

final = DiscountProduct(name, float(price), float(percent))

print(f'{final.final_price():.2f}')
```

Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(),

value().

Sau đó thực hiện dãy thao tác trên đối tượng.

Input

Dòng 1: n

Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách.

Output

Giá trị cuối cùng của count. 


```python
class Counter:
    def __init__(self):
        self.count = 0
    def inc(self):
        self.count += 1
    def dec(self):
        self.count -= 1
    def value(self):
        return self.count
n = int(input())
cmd = input().split()
c = Counter()
for i in cmd:
    if i == 'inc':
        c.inc()
    elif i == 'dec':
        c.dec()
print(c.value())
```

Yêu cầu

Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x),

pop(), top().

Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ.

Input

Dòng 1: n

n dòng tiếp theo là một trong ba lệnh:

PUSH x
POP

TOP

Output

Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng.

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, x):
        self.stack.append(x) 

    def pop(self):
        return self.stack.pop() if self.stack else None

    def top(self):
        return self.stack[-1] if self.stack else None


stack = Stack()

n = int(input())

for _ in range(n):
    cmd = input().split()

    if cmd[0] == 'PUSH':
        stack.push(int(cmd[1]))

    elif cmd[0] == 'POP':
        stack.pop()

    elif cmd[0] == 'TOP':
        print(stack.top())
```
