Yêu cầu
Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14.
Input
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
Dòng 2: tham số tương ứng.
Output
In ra diện tích với 2 chữ số thập phân.

```python
class Shape:
    def area(self):
        return 0
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r
shapes = input()
if shapes == 'rectangle':
    w, h = map(float, input().split())
    shape = Rectangle(w, h)
elif shapes == 'circle':
    r = float(input())
    shape = Circle(r)
print(f'{shape.area():.2f}')
```

Yêu cầu

Xây dựng lớp Vehicle(brand, speed).

Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.

Viết phương thức info() cho mỗi lớp:
- Vehicle: brand-speed
- Car: brand-speed-seats

Input

Dòng 1: loại đối tượng (Vehicle hoặc Car)

Dòng 2: dữ liệu tương ứng.

Output

In ra chuỗi info(). 

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def info(self):
        return f'{self.brand} - {self.speed} km/h'
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
    def info(self):
        return f'{self.brand} - {self.speed} km/h - {self.seats} seats'
obj = input()
if obj == 'vehicle':
    brand, speed = input().split()
    obj = Vehicle(brand, speed)
elif obj == 'car':
    brand, speed, seats = input().split()
    obj = Car(brand, speed, seats)
print(obj.info())

```

Xây dựng lớp Polygon với phương thức perimeter().

Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.

Input

Một dòng chứa 3 số nguyên a b c.

Output

In ra chu vi tam giác. 

```python
class Polygon:
    def perimeter(self):
        return 0
class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def perimeter(self):
        return self.a + self.b + self.c
a, b, c = map(int, input().split())
triangle = Triangle(a, b, c)
print(triangle.perimeter())
   

```

Xây dựng lớp Person(name, age) và lớp Student(name, age, school) kế thừa từ Person.

Phương thức describe():

- Person: name age

- Student: name age school

Input

Dòng 1: Person hoặc Student

Dòng 2: dữ liệu tương ứng.

Output

In ra kết quả describe(). 


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def describe(self):
        return f'{self.name} {self.age}'
class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school
    def describe(self):
        return f'{self.name} {self.age} {self.school}'
obj = input()
if obj == 'Person':
    name, age = input().split()
    obj = Person(name, age)
elif obj == 'Student':
    name, age, school = input().split()
    obj = Student(name, age, school)
print(obj.describe())
```

Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance().

Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức

add_interest() để cộng balance*rate vào số dư.

Input

balance rate

Output

Số dư sau khi cộng lãi, in 2 chữ số thập phân.

```python
class Accout:
    def __init__(balance):
        self.balance = balance
    def deposit(self, x):
        self.balance += x
    def get_balance(self):
        return self.balance
class SavingsAccount(Accout):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
    def add_interest(self):
        self.balance += self.balance * self.rate
balance, rate = map(float, input().split())
acc = SavingsAccount(balance, rate)
acc.add_interest()
print(f'{acc.get_balance():.2f}')
```

Xây dựng lớp Printer với phương thức print_value(x).

Xây dựng lớp BinaryPrinter kế thừa từ Printer, ghi đè print_value(x) để in biểu diễn nhị phân của số nguyên dương x.

Input

Một số nguyên dương n.

Output

In ra dạng nhị phân của n. 


```python
class Printer:
    def print_value(self, x):
        print(x)
class BinaryPrinter(Printer):
    def print_value(self, x):
        print(bin(x)[2:])
obj = int(input())
printer = BinaryPrinter()
printer.print_value(obj)

```

Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:

[a b]

[c d]

Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.

Input

Một dòng chứa 4 số nguyên a b c d.

Output

In ra trace và determinant trên cùng một dòng. 

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def trace(self):
        return self.a + self.d
    def determinant(self):
        return self.a * self.d - self.b * self.c
a, b, c, d = map(int, input().split())
matrix = Matrix2x2(a, b, c, d)
print(matrix.trace(), matrix.determinant())
```
