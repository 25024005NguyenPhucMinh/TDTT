Yêu cầu
Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14.
Input
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
Dòng 2: tham số tương ứng.
Output
In ra diện tích với 2 chữ số thập phân.

```python
class Shape:
    def area(self):
        return 0
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r * self.r
shapes = input()
if shapes == 'rectangle':
    w, h = map(float, input().split())
    shape = Rectangle(w, h)
elif shapes == 'circle':
    r = float(input())
    shape = Circle(r)
print(f'{shape.area():.2f}')
```

W8A17

Yêu cầu

Xây dựng lớp Vehicle(brand, speed).

Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.

Viết phương thức info() cho mỗi lớp:
- Vehicle: brand-speed
- Car: brand-speed-seats

Input

Dòng 1: loại đối tượng (Vehicle hoặc Car)

Dòng 2: dữ liệu tương ứng.

Output

In ra chuỗi info(). 

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def info(self):
        return f'{self.brand} - {self.speed} km/h'
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats
    def info(self):
        return f'{self.brand} - {self.speed} km/h - {self.seats} seats'
obj = input()
if obj == 'vehicle':
    brand, speed = input().split()
    obj = Vehicle(brand, speed)
elif obj == 'car':
    brand, speed, seats = input().split()
    obj = Car(brand, speed, seats)
print(obj.info())

```
