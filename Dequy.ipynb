đệ quy chuyển từ int -> bin

```python
def binary(a: int):
    if a <= 1:
        print(a, end="")
        return 
    binary(a // 2)
    print(a % 2, end="")
a = input()
binary(int(a))
```

đệ quy tìm số fibonacci thứ n


```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
n = int(input())
print(fibonacci(n-1))
```

merge sort

```python
def Merge(lst, start, mid, end):
    temp = []
    i = start
    j = mid

    while i < mid and j < end:
        if lst[i] < lst[j]:
            temp.append(lst[i])
            i += 1
        else:
            temp.append(lst[j])
            j += 1

    while i < mid:
        temp.append(lst[i])
        i += 1

    while j < end:
        temp.append(lst[j])
        j += 1

    for k in range(len(temp)):
        lst[start + k] = temp[k]


def MergeSort(lst, start, end):
    if end - start <= 1:
        return

    mid = (start + end) // 2

    MergeSort(lst, start, mid)
    MergeSort(lst, mid, end)

    Merge(lst, start, mid, end)


lst = [3, 2, 5, 6, 8, 1, 0, 10, 4, 7]

MergeSort(lst, 0, len(lst))
print(lst)
```
