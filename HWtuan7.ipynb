Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp


```python
lst = list(map(int, input().split()))

def beautiful(lst):
    s = 0

    for i in set(lst):
        c = lst.count(i)

        if c > i:
            s += c - i
        elif c < i:
            s += c

    return s

print(beautiful(lst))


        
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
lst = list(map(int, input().split()))

candidate = 0
count = 0

for i in lst:
    if count == 0:
        candidate = i

    if i == candidate:
        count += 1
    else:
        count -= 1

if lst.count(candidate) > len(lst) // 2:
    print(candidate)
```

Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào


```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

i = j = 0
c = []

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        c.append(a[i])
        i += 1
    else:
        c.append(b[j])
        j += 1

while i < len(a):
    c.append(a[i])
    i += 1

while j < len(b):
    c.append(b[j])
    j += 1

print(c)
```

W7A22

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())

d = {}

for i in range(n):
    s = input().split()

    ten = s[-1]   

    if ten in d:
        d[ten] += 1
    else:
        d[ten] = 1

maxx = max(d.values())

kq = []

for ten, sl in d.items():
    if sl == maxx:
        kq.append(ten)

print(min(kq))
```

Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các
số đó. 


```python
s = input()

num = ""
d = []

for i in s:
    if i.isdigit():
        num += i
    else:
        if num != "":
            d.append(int(num))
            num = ""

if num != "":
    d.append(int(num))

tong = sum(d)

if len(d) > 0:
    tb = tong / len(d)
else:
    tb = 0

print(tong)
print(tb)
```

W7A27

Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
s = input()

stack = []
ok = True

for i in s:

    if i in "([{":
        stack.append(i)

    else:
        if len(stack) == 0:
            ok = False
            break

        top = stack.pop()

        if (top == "(" and i != ")") or \
           (top == "[" and i != "]") or \
           (top == "{" and i != "}"):
            ok = False
            break

if len(stack) != 0:
    ok = False

print(ok)
```

Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
N = int(input())

def prime(x):
    if x < 2:
        return False

    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False

    return True

d = []
i = N + 1

while len(d) < 10:
    if not prime(i):
        d.append(i)
    else:
        break

    i += 1

print(d)
```

W7A29

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
lst = list(map(int, input().split()))

best = []
cur = [lst[0]]

for i in range(1, len(lst)):

    if lst[i] > lst[i - 1]:
        cur.append(lst[i])
    else:
        if len(cur) > len(best):
            best = cur

        cur = [lst[i]]

if len(cur) > len(best):
    best = cur

print(best)
```

W7A30

Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
lst = list(map(int, input().split()))

n = len(lst)

dp = [1] * n

for i in range(n):
    for j in range(i):
        if lst[i] > lst[j]:
            dp[i] = max(dp[i], dp[j] + 1)

print(max(dp))
```
