W7A1

Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):



```python
arr = list(map(int, input().split()))
target = int(input())
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
print(binary_search(arr, 0, len(arr) - 1, target))
```

W7A3

Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection
Sort)


```python
n = list(map(int, input().split()))
def bubble_sort(n):
    N = len(n)
    for i in range(N):
        swap = False
        for j in range(0, N-i-1):
            if n[j] > n[j+1]:
                n[j], n[j+1] = n[j+1], n[j]
                swap = True
        if not swap:
            break
    return n
print(bubble_sort(n))
```

```python
n = list(map(int, input().split()))
def selection_sort(n):
    N = len(n)
    for i in range(N):
        for j in range(i+1, N):
            if n[j] < n[i]:
                n[i], n[j] = n[j], n[i]
    return n
print(selection_sort(n))
```

W7A8

Bạn được cung cấp: một
danh sách các khoảng thời
gian interval = [[l1, r1], [l2,
r2], ...] và một danh sách
các truy vấn queries = [q1,
q2, ...]
Với mỗi truy vấn q, hãy tìm
độ dài nhỏ nhất của
khoảng thời gian [li, ri] sao
cho li ≤ q ≤ ri.
Nếu không tồn tại khoảng
thời gian nào như vậy, hãy
trả về -1 cho truy vấn đó.
solve in < O(m.n), hint:
binary search in the
queries list


```python
intervals = eval(input())
queries = eval(input())

ans = []

for q in queries:

    mn = float('inf')

    for l, r in intervals:

        if l <= q <= r:

            mn = min(mn, r - l + 1)

    if mn == float('inf'):
        ans.append(-1)
    else:
        ans.append(mn)

print(*ans)
```

W7A9

Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số
đếm.
Efficient solution: using
heap or binary search tree
+

```python
class Node:

    def __init__(self, val):

        self.val = val
        self.left = None
        self.right = None

        self.left_count = 0
        self.dup = 1


def insert(root, x):

    count = 0
    cur = root

    while True:

        if x == cur.val:

            cur.dup += 1
            count += cur.left_count
            break

        elif x < cur.val:

            cur.left_count += 1

            if cur.left is None:
                cur.left = Node(x)
                break

            cur = cur.left

        else:

            count += cur.left_count + cur.dup

            if cur.right is None:
                cur.right = Node(x)
                break

            cur = cur.right

    return count


nums = eval(input())

res = [0] * len(nums)

root = Node(nums[-1])

for i in range(len(nums) - 2, -1, -1):

    res[i] = insert(root, nums[i])

print(res)
```

W7A10

Cho một danh sách các số
nguyên nums và một số
nguyên m, hãy tìm tổng
mảng con lớn nhất theo
modulo m.
Tức là, với bất kỳ mảng
con nums[i:j] nào, hãy tính
sum(nums[i:j]) % m và trả
về kết quả lớn nhất có thể.
Gợi ý:
Duy trì tổng tiền tố (prefix
sum) mod m và sử dụng
tìm kiếm nhị phân để tìm
tổng tiền tố nhỏ nhất chỉ
lớn hơn tổng hiện tại một
chút (để tối đa hóa hiệu số
bao quanh).


```python
nums = eval(input())
m = int(input())

def maximum_sum(nums, m):
    prefix = 0
    ans = 0
    prefixes = []

    for x in nums:
        prefix = (prefix + x) % m

        if prefix > ans:
            ans = prefix

        pos = len(prefixes)

        for i in range(len(prefixes)):
            if prefixes[i] > prefix:
                pos = i
                break

        if pos < len(prefixes):
            value = (prefix - prefixes[pos] + m) % m

            if value > ans:
                ans = value

        inserted = False

        for i in range(len(prefixes)):
            if prefix < prefixes[i]:
                prefixes.insert(i, prefix)
                inserted = True
                break

        if not inserted:
            prefixes.append(prefix)

    return ans

print(maximum_sum(nums, m))
```

Dãy hài hòa là một dãy
thỏa mãn điều kiện giá trị
lớn nhất và giá trị nhỏ nhất
của dãy khác nhau đúng
bằng \( 1 \).
Viết hàm trả về độ dài của
dãy hài hòa con dài nhất có
thể có trong mảng giá trị
nguyên \( arr \) có \( n \)
phần tử
Ví dụ:
Input: 1 3 2 2 5 2 3 7
Output: 5 - Dãy hài hòa con
dài nhất là [3,2,2,2,3]


```python
arr = list(map(int, input().split()))

d = {}

for i in arr:
    d[i] = d.get(i, 0) + 1

ans = 0

for i in d:
    if i + 1 in d:
        ans = max(ans, d[i] + d[i + 1])

print(ans)
```
