W7A1

Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):



```python
arr = list(map(int, input().split()))
target = int(input())
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
print(binary_search(arr, 0, len(arr) - 1, target))
```

W7A2

Cho danh sách arr và một
số nguyên x, đếm xem x
xuất hiện bao nhiêu lần
trong danh sách. def
count_occurrences(arr, x) -
> int:

```python
arr = list(map(int, input().split()))
x = int(input())
def count_occurences(arr, x):
    count = 0
    for i in arr:
        if i == x:
            count += 1
    return count
print(count_occurences(arr, x))
```

W7A3

Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection
Sort)


```python
n = list(map(int, input().split()))
def bubble_sort(n):
    N = len(n)
    for i in range(N):
        swap = False
        for j in range(0, N-i-1):
            if n[j] > n[j+1]:
                n[j], n[j+1] = n[j+1], n[j]
                swap = True
        if not swap:
            break
    return n
print(bubble_sort(n))
```

```python
n = list(map(int, input().split()))
def selection_sort(n):
    N = len(n)
    for i in range(N):
        for j in range(i+1, N):
            if n[j] < n[i]:
                n[i], n[j] = n[j], n[i]
    return n
print(selection_sort(n))
```

W7A4

Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.


```python
n = list(map(int, input().split()))
def count(n):
    most_frequent = None
    max_count = 0
    for i in n:
        count = n.count(i)
        if count > max_count:
            max_count = count
            most_frequent = i
            if i < most_frequent:
                most_frequent = i
    return f'{most_frequent} xuat hien nhieu nhat, som nhat, {max_count} lan'
print(count(n))
```

W7A7

Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền kề).


```python
lst = list(map(int, input().split()))
dp = [1] * len(lst)
for i in range(len(lst)):
    for j in range(i + 1, len(lst)):
        if lst[j] > lst[i]:
            dp[j] = max(dp[j], dp[i] + 1)
print(max(dp))
```

W7A8

Bạn được cung cấp: một
danh sách các khoảng thời
gian interval = [[l1, r1], [l2,
r2], ...] và một danh sách
các truy vấn queries = [q1,
q2, ...]
Với mỗi truy vấn q, hãy tìm
độ dài nhỏ nhất của
khoảng thời gian [li, ri] sao
cho li ≤ q ≤ ri.
Nếu không tồn tại khoảng
thời gian nào như vậy, hãy
trả về -1 cho truy vấn đó.
solve in < O(m.n), hint:
binary search in the
queries list


```python
intervals = eval(input())
queries = list(map(int, input().split()))

ans = []

for q in queries:

    mn = float('inf')

    for l, r in intervals:

        if l <= q <= r:

            mn = min(mn, r - l + 1)

    if mn == float('inf'):
        ans.append(-1)
    else:
        ans.append(mn)

print(*ans)
```

W7A9

Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số
đếm.
Efficient solution: using
heap or binary search tree
+

```python
class Node:

    def __init__(self, val):

        self.val = val
        self.left = None
        self.right = None

        self.left_count = 0
        self.dup = 1


def insert(root, x):

    count = 0
    cur = root

    while True:

        if x == cur.val:

            cur.dup += 1
            count += cur.left_count
            break

        elif x < cur.val:

            cur.left_count += 1

            if cur.left is None:
                cur.left = Node(x)
                break

            cur = cur.left

        else:

            count += cur.left_count + cur.dup

            if cur.right is None:
                cur.right = Node(x)
                break

            cur = cur.right

    return count


nums = list(map(int, input().split()))

res = [0] * len(nums)

root = Node(nums[-1])

for i in range(len(nums) - 2, -1, -1):

    res[i] = insert(root, nums[i])

print(res)
```
