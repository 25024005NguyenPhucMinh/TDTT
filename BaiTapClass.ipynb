Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao

(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện

tích toàn phần) và thể tích của hình trụ

Đầu vào

Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và

chiều cao của hình trụ, cách nhau bởi một dấu cách.

Đầu ra

In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử

dụng lại những hàm vừa viết, π=3.14).

(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 

```python
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h
    def surface_area(self) -> float:
        return 2 * 3.14 * self.r * self.h + 2 * 3.14 * self.r**2
    def volume(self) -> float:
        return 3.14 * self.r**2 * self.h
rad, height = map(float, input().split())
cyl = Cylinder(rad, height)
print(f'{cyl.surface_area():.2f} {cyl.volume():.2f}')
```

[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương

thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không

và phương thức tính ngày kế tiếp của ngày tháng năm nào đó

nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn

dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in

ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.

Đầu vào

Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng

dd/mm/yyyy
.

Đầu ra

In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng

dd/mm/yyyy

. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class Date:
    def __init__(self, dd, mm, yy):
        self.d = dd
        self.m = mm
        self.y = yy
    def checkValid(self) -> bool:
        maxDay = -1
        match self.m:
            case 1 | 3 | 5 | 7 | 8 | 10 | 12:
                maxDay = 31
            case 4 | 6 | 9 | 11:
                maxDay = 30
            case 2:
                if self.y % 4 == 0 and (self.y % 100 != 0 or self.y % 400 == 0):
                    maxDay = 29
                else:
                    maxDay = 28
        return self.y > 0 and 1 <= self.d <= maxDay
    def getnextDay(self):
        nextday = Date(self.d + 1, self.m, self.y)
        if nextday.checkValid():
            return nextday
        nextday.d = 1
        nextday.m += 1
        if nextday.checkValid():
            return nextday
        nextday.m = 1
        nextday.y += 1
        return nextday
d, m, y = map(int, input().split('/'))
date = Date(d, m, y)
if date.checkValid():
    nextday = date.getnextDay()
    print(f'{nextday.d}/{nextday.m}/{nextday.y}')
else:
    print('Invalid')
```

[ShoppingCart]
Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản
phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi
sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:

1. Khởi tạo với danh sách rỗng

2. Thêm vào giỏ hàng một sản phẩm mới

3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)

4. Kiểm tra xem giỏ hàng có rỗng không

5. Tính tổng tiền giỏ hàng

6. Hiển thị toàn bộ sản phẩm trong giỏ

7. Xóa toàn bộ giỏ hàng

Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có
muốn thoát khỏi chương trình không?"

```python
class ShoppingCart:
    def __init__(self):
        self.cart = []
    def add_item(self, name, price):
        self.cart.append({'name': name, 'price': float(price)})
    def remove_item(self, name):
        for product in self.cart:
            if product['name'] == name:
                self.cart.remove(product)
                return f'item {name} removed'
        return f'item {name} not found'
    def is_empty(self):
        return len(self.cart) == 0
    def total_price(self):
        total = 0
        for product in self.cart:
            total += product['price']
        return total
    def show_cart(self):
        if self.is_empty():
            return 'Cart is empty'
        else:
            for product in self.cart:
                print(f"{product['name']}: {product['price']:.2f}")
    def clear_cart(self):
        self.cart.clear()
        return 'Cart cleared'
cart = ShoppingCart()
while True:
    print('\n2. Add product')
    print('3. Remove product')
    print('4. Show cart')
    print('5. Total price')
    print('6. Show cart')
    print('7. Clear cart')

    choice = input()
    if choice == '2':
        name = input('Enter product name: ')
        price = float(input('Enter product price: '))
        cart.add_item(name, price)
    elif choice == '3':
        name = input('Enter product name to remove: ')
        print(cart.remove_item(name))
    elif choice == '4':
        if cart.is_empty():
            print('Cart is empty')
        else:
            print('Cart is not empty')
    elif choice == '5':
        print(f'Total price: {cart.total_price()}')
    elif choice == '6':
        cart.show_cart()
    elif choice == '7':
        print(cart.clear_cart())
    else:
        print('Invalid choice')
    stop = input('Do you want to exit? ')
    if stop in ['yes', 'YES', 'Yes']:
        break
```

Input

w h k (số thực dương)

```python
class Rectangle:
    def __init__(self, ww, hh):
        self.w = ww
        self.h = hh
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self, k):
        self.w *= k
        self.h *= k
w, h, k = map(float, input().split())
rect = Rectangle(w, h)
rect.scale(k)
print(f'Area: {rect.area():.2f} Perimeter: {rect.perimeter():.2f}')
```

Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:

Rút gọn phân số trong __init__ (quy ước mẫu > 0).

Các phương thức add/sub/mul/div trả về phân số đã rút gọn.

__str__() in dạng a/b.

Input

a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})

Output

Phân số kết quả đã rút gọn 


```python
from math import gcd

class Fraction:
    def __init__(self, num, den):
        g = gcd(num, den)

        num //= g
        den //= g

        if den < 0:
            num *= -1
            den *= -1

        self.num = num
        self.den = den

    def add(self, other):
        return Fraction(
            self.num * other.den + other.num * self.den,
            self.den * other.den
        )

    def sub(self, other):
        return Fraction(
            self.num * other.den - other.num * self.den,
            self.den * other.den
)

    def mul(self, other):
        return Fraction(
            self.num * other.num,
            self.den * other.den
        )

    def div(self, other):
        return Fraction(
            self.num * other.den,
            self.den * other.num
        )

    def __str__(self):
        return f"{self.num}/{self.den}"


a, b, op, c, d = input().split()

a = int(a)
b = int(b)
c = int(c)
d = int(d)

f1 = Fraction(a, b)
f2 = Fraction(c, d)

if op == '+':
    result = f1.add(f2)

elif op == '-':
    result = f1.sub(f2)

elif op == '*':
    result = f1.mul(f2)

elif op == '/':
    result = f1.div(f2)

print(result)
```

Lớp BankAccount(owner, balance) với:

Input

Dòng 1: owner initial_balance

Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.

Output

Số dư cuối cùng (in số thực, 2 chữ số). 



```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def deposit(self, x):
        self.balance += x
        return self.balance
    def withdraw(self, x):
        if self.balance >= x:
            self.balance -= x
            return self.balance
        else:
            print('Not enough balance')
    def __str__(self):
        return f'Owner: {self.owner}, Balance: {self.balance:.2f}'
ow, ba = input().split()
ba = float(ba)
account = BankAccount(ow, ba)

n = int(input())
for _ in range(n):
    op, x = input().split()
    x = float(x)
    if op == 'DEPOSIT':
        account.deposit(x)
    elif op == 'WITHDRAW':
        account.withdraw(x)
print(f'{account.balance:.2f}')
```

Input

Dòng 1: name

Dòng 2: n

n dòng tiếp: subject score (score là số thực 0–10)

Output

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = {}
    def add_score(self, subject, score):
        self.scores[subject] = score
    def avg(self):
        return sum(self.scores.values()) / len(self.scores)
    def rank(self):
        average = self.avg()
        if average >= 8:
            return "Excellent"
        elif average >= 6.5:
            return "Good"
        elif average >= 5:
            return "Average"
        else:
            return "Poor"
name = input()
student = Student(name)
n = int(input())
for _ in range(n):
    sub, sco = input().split()
    sco = float(sco)
    student.add_score(sub, sco)
print(f'{student.name} {student.avg():.2f} {student.rank()}')
        
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
class Library:
    def __init__(self):
        self.books = []
    def add_book(self, book):
        self.books.append(book)
    def count_by_author(self, name):
        count = 0
        for book in self.books:
            if book.author == name:
                count += 1
        return count
    def find_by_year(self, y):
        count = 0
        for book in self.books:
            if book.year == y:
                count += 1
        return count
library = Library()
n = int(input())
for _ in range(n):
    s = input()
    if s.startswith("ADD"):
        data = s[4:].split(";")

        title = data[0]
        author = data[1]
        year = int(data[2])

        book = Book(title, author, year)

        library.add(book)

    elif s.startswith("COUNTYEAR"):
        year = int(s.split()[1])

        print(library.find_by_year(year))

    elif s.startswith("COUNT"):
        author = s[6:]

        print(library.count_by_author(author))

```

Bạn được cho bốn điểm A, B, C và D trong hệ tọa độ Đề-các 3
chiều.

Nhiệm vụ của bạn là tính và in ra góc giữa hai mặt phẳng tạo
bởi các điểm:

mặt phẳng (A, B, C)

mặt phẳng (B, C, D)

Góc cần in ra được ký hiệu là PHI, và phải dùng độ (degrees),
không dùng radian.

Dạng Input:

Một dòng input chứa 3 số thực, ngăn cách bằng dấu cách, biểu
diễn:

tọa độ X tọa độ Y tọa độ Z của một điểm.

Dạng Output:

In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập
phân

```python
import math


def vector(a, b):
    return (
        b[0] - a[0],
        b[1] - a[1],
        b[2] - a[2]
    )


def cross(u, v):
    return (
        u[1] * v[2] - u[2] * v[1],
        u[2] * v[0] - u[0] * v[2],
        u[0] * v[1] - u[1] * v[0]
    )


def dot(u, v):
    return u[0]*v[0] + u[1]*v[1] + u[2]*v[2]


def norm(v):
    return math.sqrt(dot(v, v))


A = tuple(map(float, input().split()))
B = tuple(map(float, input().split()))
C = tuple(map(float, input().split()))
D = tuple(map(float, input().split()))

AB = vector(A, B)
AC = vector(A, C)

BC = vector(B, C)
BD = vector(B, D)


n1 = cross(AB, AC)
n2 = cross(BC, BD)

cos_theta = abs(dot(n1, n2)) / (norm(n1) * norm(n2))


cos_theta = min(1, max(-1, cos_theta))

theta = math.degrees(math.acos(cos_theta))

print(f"{theta:.2f}")
```

[Complex]
Khai báo kiểu dữ liệu biểu diễn số phức \(ComplexNumber\)
bao gồm \(2\) biến \(realPart\) và \(imaginaryPart\) kiểu
\(double\) biểu diễn phần thực và phần ảo của một số phức.
Viết hàm khởi tạo ComplexNumber(double initRealPart, double
initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo
của một số phức và hàm ComplexNumber
multiply(complexNumber other) trả về tích của số phức hiện tại
và một số phức khác.
Đầu vào
Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực
cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo
của một số phức. Các số trong đầu vào có giá trị tuyệt đối
không vượt quá \(100\).
Đầu ra
In ra màn hìn

```python
class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.realPart = initRealPart
        self.imaginaryPart = initImaginaryPart

    def multiply(self, other):
        real = (self.realPart * other.realPart
                - self.imaginaryPart * other.imaginaryPart)

        imag = (self.realPart * other.imaginaryPart
                + self.imaginaryPart * other.realPart)

        return ComplexNumber(real, imag)


a, b = map(float, input().split())
c, d = map(float, input().split())

z1 = ComplexNumber(a, b)
z2 = ComplexNumber(c, d)

result = z1.multiply(z2)

print(f"{result.realPart:.2f} {result.imaginaryPart:.2f}")
```

[Triangle]
Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa \(3\) biến
\(egde1,\,edge2,\,edge3\) kiểu \(int\) là độ dài \(3\) cạnh của
tam giác đó.
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int
initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm
double getArea() trả về diện tích của tam giác đó (nếu tam giác
không hợp lệ, trả về \(-1\)).
Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một
tam giác và in ra diện tích của tam giác đó.
Đầu vào
Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài
\(3\) cạnh của tam giác
Đầu ra
In ra màn hình diện tích tam giác (làm tròn đến\(2\) chữ số thập
phân sau dấu phẩy). In ra \(invalid\) nếu \(3\) số nhập vào
không phải là \(3\) cạnh của một tam giác 


```python
import math


class Triangle:
    def __init__(self, initEdge1, initEdge2, initEdge3):
        self.edge1 = initEdge1
        self.edge2 = initEdge2
        self.edge3 = initEdge3

    def getArea(self):
        a = self.edge1
        b = self.edge2
        c = self.edge3

        if a + b <= c or a + c <= b or b + c <= a:
            return -1

        p = (a + b + c) / 2

        return math.sqrt(p * (p - a) * (p - b) * (p - c))


a, b, c = map(int, input().split())

triangle = Triangle(a, b, c)

area = triangle.getArea()

if area == -1:
    print("invalid")
else:
    print(f"{area:.2f}")
```

Yêu cầu

Xây dựng lớp Employee(name, salary).

Xây dựng lớp Manager(name, salary, bonus) kế thừa từ
Employee.

Cả hai lớp đều có phương thức income():

- Employee: trả về salary

- Manager: trả về salary + bonus

Input

Dòng 1: loại đối tượng (Employee hoặc Manager)

Dòng 2: name salary [bonus]

Output

In ra tổng thu nhập của đối tượng. 

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def income(self):
        return self.salary
class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus
    def income(self):
        return self.salary + self.bonus
obj = input()
if obj == "Employee":
    name, salary = input().split()
    salary = float(salary)
    employee = Employee(name, salary)
    print(f'{employee.name} {employee.income():.2f}')
elif obj == "Manager":
    name, salary, bonus = input().split()
    salary = float(salary)
    bonus = float(bonus)
    manager = Manager(name, salary, bonus)
    print(f'{manager.name} {manager.income():.2f}')
```
